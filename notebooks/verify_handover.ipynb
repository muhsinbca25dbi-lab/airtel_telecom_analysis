```python
pq("""select customer_id, count(*) as orders, sum(case when amount >= 500 then 1 else 0 end) as greater_500 from shop_orders group by customer_id;""")
```

```python
pq("""select count(*) as orders, sum(amount >= 500) as big_orders, round(100.0*avg(amount >= 500),2) as perc from shop_orders;""")
```

```python
pq("""select name, joined_on, cast(julianday('2025-03-31')-julianday(joined_on) as int) as days from shop_customers;""")
```

```python
pq("""select c.name, c.city from shop_customers as c left join shop_orders as o on o.customer_id = c.customer_id where o.customer_id is null;""")
```
